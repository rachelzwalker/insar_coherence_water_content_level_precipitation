### Set area of interest, timing and other requirements
You will need to get the aoi and relative orbit by looking on the ASF website: https://search.asf.alaska.edu/#/ (draw a polygon on the map and get the aoi from the Area of Interest WKT section of the bar and click 'SEARCH' to get scene details. The relative orbit is the same as the 'Path'

```python
aoi = 'POLYGON((-2.0395 53.5805,-2.0389 53.4645,-1.8289 53.4657,-1.8305 53.5834,-2.0395 53.5805))'
start = '2019-12-26'
end = '2019-12-31'

opts = {
    'platform': asf.PLATFORM.SENTINEL1,
    'start': start,
    'end': end,
    'beamMode': asf.BEAMMODE.IW,
    'flightDirection': asf.FLIGHT_DIRECTION.ASCENDING,
    'relativeOrbit': 132
}
```

#### Set the reference if you know it, otherwise, granule_1=None and run the following cell which will give you a list of slc granules in the timeframe you have given. Choose the reference based on the date e.g. below date is 20180402.

```python
granule_1 = 'S1B_IW_SLC__1SDV_20180402T174910_20180402T174937_010308_012C0D_BC27'
```

```python
if granule_1 == None:
    results = asf.geo_search(intersectsWith=aoi, **opts)
    granules = []
    for result in range(len(results)):
        granules.append(results[result].properties['sceneName'])
    granules.reverse() # otherwise list has the first one at the end
    granules = list(dict.fromkeys(granules))
    granules_slc = [x for x in granules if "SLC" in x]
    granules_slc_focused = [x for x in granules_slc if "OPERA" not in x]

    print(granules_slc_focused)
```

### Set save directory

```python
current_directory = os.getcwd()
save_directory = f"{current_directory}/resources"
```

### Get the data

```python
DataRetrieval(aoi, opts, save_directory, granule_1).retrieve_asf_data()
```
